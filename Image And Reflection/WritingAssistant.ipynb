```python
topic = input("Enter TOEFL writing topic: ")
```

```python
from google.colab import userdata
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

api_key = userdata.get("OpenRouter")

writer_llm = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=api_key,
    base_url="https://openrouter.ai/api/v1",
    temperature=0.7
)

reflector_llm = ChatOpenAI(
    model="openai/gpt-4o-mini",
    api_key=api_key,
    base_url="https://openrouter.ai/api/v1",
    temperature=0.2
)
```

```python
writer_system_prompt = """
You are a TOEFL writing student.

Write an English essay in response to the given TOEFL topic.

Requirements:
- Write like a real TOEFL candidate.
- Clearly state your opinion.
- Use an introduction, body paragraphs, and conclusion.
- Give reasons and examples.
- Use natural academic English.
- Do not make the essay unrealistically perfect.
"""

writer_messages = [
    SystemMessage(content=writer_system_prompt),
    HumanMessage(content=topic)
]

writer_response = writer_llm.invoke(writer_messages)

first_essay = writer_response.content

print("===== FIRST ESSAY =====")
print(first_essay)
```

```python
reflector_system_prompt = """
You are a TOEFL writing evaluator.

Evaluate the student's essay using these criteria:

1. Coherence
2. Language accuracy
3. Structure
4. Vocabulary

Your response must contain:

Score: a score from 0 to 5

Strengths:
- Explain the strengths of the essay.

Weaknesses:
- Explain the weaknesses of the essay.

Suggestions:
- Give specific suggestions for improving the essay.

Humanization:
- Suggest changes that make the essay sound more natural
  and less machine-generated or artificial.

Do NOT rewrite the essay.
Only evaluate it and provide feedback.
"""
reflector_messages = [
    SystemMessage(content=reflector_system_prompt),
    HumanMessage(
        content=f"""
TOEFL Topic:
{topic}

Student Essay:
{first_essay}
"""
    )
]

reflection_response = reflector_llm.invoke(reflector_messages)

feedback = reflection_response.content

print("===== REFLECTOR FEEDBACK =====")
print(feedback)
```

```python
writer_messages.append(AIMessage(content=first_essay))
writer_messages.append(
    AIMessage(content=first_essay)
)

writer_messages.append(
    HumanMessage(
        content=f"""
A TOEFL evaluator reviewed your essay and provided the following feedback:

{feedback}

Rewrite your original essay based on this feedback.

Requirements:
- Fix the weaknesses identified by the evaluator.
- Keep the strong parts of the original essay.
- Improve coherence, grammar, structure, and vocabulary.
- Make the writing sound natural and human.
- Keep the same position and main ideas.
- Return only the revised essay.
"""
    ))
writer_response = writer_llm.invoke(writer_messages)

second_essay = writer_response.content

print("===== SECOND ESSAY =====")
print(second_essay)
```
